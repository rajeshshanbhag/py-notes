```python
import numpy as np

my_array = np.array([1, 2, 3, 4, 5])
print(my_array)
my_slice = my_array[1:3]
my_slice
```

```python
my_array = np.array([1, 2, 3, 4, 5])
my_slice = my_array[1:3]
my_array = my_array * 2
my_slice
```

```python
my_array = np.array([1, 2, 3, 4, 5])
my_slice = my_array[1:3]
my_array[:] = my_array * 2
my_slice
```

```python
my_array = np.array([1, 2, 3, 4, 5])
my_slice = my_array[1:3].copy()
my_array[:] = my_array * 2
my_slice
```

```python
my_array = np.array([1, 2, 3, 4, 5])
my_slice = my_array[[1, 2]]
my_array[:] = my_array * 2
my_slice
```

```python
my_array = np.array([1, 2, 3, 4, 5])
my_slice = my_array[1:3]
my_slice[0] = -99
my_array
```

```python
my_array = np.array([1, 2, 3, 4, 5])
my_slice = my_array[1:3]
my_array = my_array * 2
my_slice[0] = -99
my_array
```
