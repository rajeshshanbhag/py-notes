```python
#### Random numbers on a histogram
```

```python
# Generate an array of 10,000 random normal numbers
random_array = np.random.randn(10000)

# Output a histogram of the data with 10 bins
bins = np.arange(-3.5, 3.5, 0.25)
hist, bin_edges = np.histogram(random_array, bins=bins)
```

```python
import matplotlib.pyplot as plt

#the histogram outputs the bin edges for value ranges used for counting how many random numbers fall into each range, 
#so that's what we eliminate the final value of bin_edges from the plot and recenter the range of each bar by adding 0.125 to the x-axis value

plt.bar(bin_edges[:-1] + 0.125, hist)
plt.xlabel("Value")
plt.ylabel("Count")
plt.show()
```

```python
#### Random Sampling
```

```python
ids = np.arange(20) # ID numbers for each member of the population we wish to sample from
print(ids)
ids_shuffle = ids.copy()
np.random.shuffle(ids_shuffle)
print(ids_shuffle)
```

```python
control = ids_shuffle[0:10]
experiment = ids_shuffle[10:]
print(control)
print(experiment)
```
