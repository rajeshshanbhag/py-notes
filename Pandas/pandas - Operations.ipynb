#### Operations

```python
# DataFrame.aggregate(func, axis=0, *args, **kwargs)  
df.aggregate({'Age':['sum','min'], 'Salary':['max','min']})
```

```python
df.Age.min(),df.Age.max(),df.Age.mean(),df.Age.median(),df.Age.mode()
```

```python
df[df.Position=='PG'].Age.min(),df[df.Position=='PG'].Age.max(),df[df.Position=='PG'].Age.mean(),df[df.Position=='PG'].Age.median(),df[df.Position=='PG'].Age.mode()
```

#### Apply a function

```python
def func(Age):
    if Age>25:
        return "High"
    else:
        return "Low"
df['Type']=df['Age'].apply(func)
df.head(7)
```

```python
df['Type']=df['Age'].apply(lambda Age:'High' if Age>25 else 'Low')
df.head(7)
```
