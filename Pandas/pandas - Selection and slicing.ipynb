### Slicing using str.slice(start,stop,step)

```python
df["Name (int)"]= df["Name"].str.slice(0,-2,2)
print(df.head())
```

#### Slicing columns

```python
df1 = pd.DataFrame({"a": [1, 2, 3, 4, 5, 6, 7], 
                    "b": [2, 3, 4, 2, 3, 4, 5], 
                    "c": [3, 4, 5, 2, 3, 4, 5], 
                    "d": [4, 5, 6, 2, 3, 4, 5], 
                    "e": [5, 6, 7, 2, 3, 4, 5]})
print(df1)
```

```python
df2 = df1.reindex(columns = ['c','b'])
print(df2)
```

```python
#df.loc[:,first:last:step]
df2 = df1.loc[:, "a":"e":2]
print(df2)
```

```python
#df.iloc[start:stop:step]
df2 = df1.iloc[:, 1:3:1]
print(df2)
```
