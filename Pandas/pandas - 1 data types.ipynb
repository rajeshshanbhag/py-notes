```python
import pandas as pd
import numpy as np
```

#### SERIES (one dimensional array)

#### Series is an ordered collection of values, generally all the same type
#### Series are basically one-dimensional numpy arrays with lots of extra features (e.g. explicit index)

#### 1. Create Series from list

```python
a = [1, 7, 2]
ser = pd.Series(data=a,name='Head')
print(ser)                    # default labels starting index 0
```

#### 2. Create Series from numpy array

```python
data = np.array(['a','b','c', 1,2,3])
ser = pd.Series(data,name="Head")
print(ser[3:])                  # access first 3 elements from Series
```

#### Define custom index to Series

```python
a = [1, 7, 2]
ser = pd.Series(a,index=['a','b','c'])
print(ser['b'])               # Access element using index (label)
```

#### Convert datatype

```python
a = [1, 7, 2]
ser = pd.Series(a,index=['a','b','c'])
ser=ser.astype('float64')
ser
```

#### 3. Create Series from key:value objects

```python
calories = {"day1": 420, "day2": 380, "day3": 390}
ser = pd.Series(calories)
print(ser)                    # keys are labels/ index
```

```python
calories = {"day1": 420, "day2": 380, "day3": 390}
ser=pd.Series(calories,index=['day1'])
print(ser)                    # keys are labels/ index
```

#### DATAFRAMES (multi dimensional array)

#### 1. Create dataframe manually

```python
data = {
  "calories": [420,380,390],
  "duration": [50,40,45]
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.loc[1])          # output is a panda series
```

```python
print(df.loc[[0,1]])
```

```python
df = pd.DataFrame(data, index = ["day1", "day2", "day3"])
print(df) 
```

#### Object datatype is slower than integer datatype

```python
import pandas as pd
import numpy as np

object_numbers = pd.Series(np.arange(1000000), dtype="object")
numbers = pd.Series(np.arange(1000000), dtype="int64")

%timeit object_numbers * 2
%timeit numbers * 2
```
