```python
import pandas as pd
pd.set_option("mode.copy_on_write", True)
```

```python
sales = pd.DataFrame(
    data={
        "employee": [
            "Katrina",
            "Guanyu",
            "Jan",
            "Roman",
            "Jacqueline",
            "Paola",
            "Esperanza",
            "Alaina",
            "Egweyn",
        ],
        "sales": [14, 17, 6, 12, 8, 3, 7, 15, 5],
        "year": [2018, 2019, 2020, 2018, 2020, 2019, 2019, 2020, 2020],
    }
)
print(sales)
```

```python
sales_by_employee = sales.groupby("employee")["sales"].sum()
sales_by_employee
```

```python
sales_by_employee.max()
```

```python
employees_by_year = sales.groupby("year")["employee"].count().reset_index()
print(employees_by_year)
```

```python
import pandas as pd

df = pd.DataFrame(data={'books':['bk1','bk1','bk1','bk2','bk2','bk3'], 'price': [12,12,12,15,15,17]})

print(df)

print(df.groupby('books', as_index=True).sum())
print(df.groupby('books').sum())
print(df.groupby('books', as_index=False).sum())
```

```python
sales_by_year_employee = sales.groupby(["year","employee"],as_index=False)["sales"].sum()
print(sales_by_year_employee)
```

#### groupby() using apply

```python
def my_sum_10_percent_less(x):
    real_sum = sum(x)
    return 0.9 * real_sum


sales_by_year = sales.groupby("year", as_index=False)["sales"].apply(
    my_sum_10_percent_less
)
print(sales_by_year)
```
