#### Operations

```python
attendance = pd.Series(
    [132, 94, 112, 84, 254, 322, 472],
    index=[
        "Monday",
        "Tuesday",
        "Wednesday",
        "Thursday",
        "Friday",
        "Saturday",
        "Sunday",
    ],
)

revenue = attendance * 15
print("revenue:",revenue)
print("Sum of revenue:",revenue.sum())
print("Power of 2:",revenue**2)
```
