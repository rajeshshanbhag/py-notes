```python
df4=df[['race','inctot','perwt','empstat','hispan']]
df4.head(10)
```

```python
df4['wt']=df4['inctot']*df4['perwt']
df4.head(10)
```

```python
df5=df4[(df4.race=='white')&(df4.empstat=='employed')]
#print(df5.head(10))
(df5.wt.sum()/df5.perwt.sum()).round(2)
```

```python
df6=df4[(df4.race=='black/african american/negro')&(df4.empstat=='employed')]
(df6.wt.sum()/df6.perwt.sum()).round(2)
```

```python
df7=df4[(df4.race=='white')&(df4.empstat=='employed')&(df4.hispan=='not hispanic')]
#print(df7.head(10))
(df7.wt.sum()/df7.perwt.sum()).round(2)
```
