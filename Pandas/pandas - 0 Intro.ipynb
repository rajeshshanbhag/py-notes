<h3 align="left" style="color:blue">pandas</h3>

##### pandas is an open-source software library built on Python for data analysis and data manipulation. The pandas library provides data structures designed specifically to handle tabular datasets with a simplified Python API. 
##### At the core of the pandas open-source library is the DataFrame data structure for handling tabular and statistical data.

##### https://www.nvidia.com/en-au/glossary/pandas-python/

#### What you can do easily with pandas that can be rather complicated without it:

1. Quickly reading your data into a structured tabular form

```python
import pandas as pd

df=pd.read_csv("smallworld.csv")
df
```

2. Quickly describe/summarize your data

```python
df.describe()
```

3. Quickly querying your dataset

```python
df.loc[df.gdp_per_capita_2008<10000]
```

4. Quickly plotting your data

```python
df.plot.bar(x="country", y="gdp_per_capita_2008")
```

#### Drawback: It is not suitable for handling 3D or N-D matrices (where N > 3). In such cases, revert to numpy or use the xarray package
