#### DATA CLEANSING

#### Drop rows with empty cells

```python
df2=df.dropna()                           # dropna() returns new dataframe
# row 18, 22, 28 deleted
#df=df.dropna(inplace=True)                  # modify original dataframe
print(df2)
```

```python
df2=df.dropna(axis=1)                        #drop columns with blank values
print(df2)
```

#### Replace all empty cells

```python
df3=df.fillna(130)
print(df3)
```

#### Replace empty cells in specific columns

```python
df4=df.fillna({"College":'not known'})
print(df4)
```

#### Replace with function value {mean(), median(), mode())

```python
df.Salary.median()
```

```python
new4_df=df.fillna({
    "Salary":df["Salary"].median()
})
print(new4_df)                                     # index 2 Salary replaced with median
```

#### Correct Date format

#### Replacing values

```python
new4_df
```

#### Replace with .loc()

```python
new4_df.loc[4, 'College'] = 'Texas'                # replace fixed value
print(new4_df)
```

#### Replace with replace() individual values

```python
new4_df['College']=new4_df['College'].replace('Texas','TX')
new4_df
```

#### Replace with replace() dictionary

```python
values={
    'Kentucky':'KT',
    'Kansas':'KN'
}
new4_df['College']=new4_df['College'].replace(values)
new4_df
```

#### Replace with replace() regex

```python
new4_df['Name']=new4_df['Name'].replace('\.','',regex=True)
new4_df
```

#### Using string methods

```python
new4_df['Name']=new4_df['Name'].str.upper()
new4_df
```

#### Identify duplicate rows

```python
df[df.duplicated()]                # identify duplicated rows
```

#### Removing duplicate rows

```python
df.drop_duplicates(inplace=True)
print(df)
```

#### Rename columns

```python
df.rename(columns={'Salary':'Sal'}, inplace=True)
print(df)
```
