```python
import pandas as pd
pd.set_option("mode.copy_on_write", True)
```

#### Concatenate

```python
hom1=pd.read_csv("homicides_1.csv")
print(hom1)
print('\n')

hom2=pd.read_csv("homicides_2.csv")
print(hom2)
print('\n')

hom=pd.concat([hom1,hom2]).reset_index(drop=True)
print(hom)
```

#### Concat does not work well with columns, use merge instead

```python
hom3=pd.concat([hom1,hom2],axis=1).reset_index(drop=True)
print(hom3)
```

#### Merge 1:1

```python
print(hom1)
print('\n')

pop1=pd.read_csv("population_1.csv")
print(pop1)
print('\n')

#x=pd.merge(a,b,how='left',on='name')
hom_pop1=pd.merge(hom1,pop1,'left','State_Code')
print(hom_pop1)
```

#### Merge 1:m

```python
a = pd.DataFrame(
    data={
        "name": ["Mia", "Lucas", "Ang", "Jia"],
        "course": ["math", "chem", "econ", "chem"],
    }
)
print(a)
print('\n')

c = pd.DataFrame(data={"course": ["chem", "econ", "math"], "credits": [4, 3, 1.5]})
print(c)
print('\n')

y=pd.merge(c,a,'left','course')
print(y)
```

```python
#### Merge m:m
```

```python
a = pd.DataFrame(
    data={
        "name": ["Mia", "Lucas", "Ang", "Jia"],
        "course": ["math", "chem", "econ", "chem"],
    }
)
print(a)
print('\n')

d = pd.DataFrame(
    data={
        "course": ["chem", "chem", "econ", "math"],
        "topic": ["atoms", "lab", "trade", "calc."],
    }
)
print(d)
print('\n')

z=pd.merge(a,d,'left','course')
print(z)
```

```python
#### Merge "left"
```

```python
p=pd.DataFrame(data={"C1": ["A", "B", "C"], "C2": ["x", "y", "z"]})
print(p)
print('\n')

q=pd.DataFrame(data={"C1": ["A", "B", "D"], "C3": [11, 12, 13]})
print(q)
print('\n')

r=pd.merge(p,q,how="left",on="C1")
print(r)
```

```python
#### Merge "right"
```

```python
p=pd.DataFrame(data={"C1": ["A", "B", "C"], "C2": ["x", "y", "z"]})
print(p)
print('\n')

q=pd.DataFrame(data={"C1": ["A", "B", "D"], "C3": [11, 12, 13]})
print(q)
print('\n')

r=pd.merge(p,q,how="right",on="C1")
print(r)
```

```python
#### Merge "inner"
```

```python
p=pd.DataFrame(data={"C1": ["A", "B", "C"], "C2": ["x", "y", "z"]})
print(p)
print('\n')

q=pd.DataFrame(data={"C1": ["A", "B", "D"], "C3": [11, 12, 13]})
print(q)
print('\n')

r=pd.merge(p,q,how="inner",on="C1")
print(r)
```

```python
#### Merge "outer"
```

```python
p=pd.DataFrame(data={"C1": ["A", "B", "C"], "C2": ["x", "y", "z"]})
print(p)
print('\n')

q=pd.DataFrame(data={"C1": ["A", "B", "D"], "C3": [11, 12, 13]})
print(q)
print('\n')

r=pd.merge(p,q,how="outer",on="C1")
print(r)
```

```python
#### Merge with indicator function
```

```python
df1 = pd.DataFrame({"key": ["key1", "key2"], "df1_var": [1, 2]})
print(df1)
print('\n')
df2 = pd.DataFrame({"key": ["key1", "Key2"], "df2_var": ["a", "b"]})
print(df2)
print('\n')
new_data = pd.merge(df1, df2, on="key", how="outer")
print(new_data)
```

```python
# indicator function adds a new column _merge
new_data = pd.merge(df1, df2, on="key", how="outer", indicator=True)
print(new_data)
print('\n')
new_data._merge.value_counts()
```

```python
#### As a good practice, use both the validate keyword and the indicator keyword (with how='outer' when possible).
```
