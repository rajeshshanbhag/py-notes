#### Copy-on-Write (CoW)

#### The idea of Copy-on-Write is that, from the perspective of the user, pandas behaves as if all operations return a copy.
####  pandas will create a view (so it's faster) and make a note that if either the original array or the subset are ever changed, it needs to actually make the copy before allowing those changes to take place

```python
import pandas as pd
pd.set_option("mode.copy_on_write", True)
```

```python
df = pd.DataFrame({"a": [10, 20, 30, 40], "b": [11, 12, 13, 14]})
print(df)
```

```python
slice = df.iloc[1:3,]                  # created as view not as a copy
print(slice)
```

```python
df.iloc[1, 1] = -1                     # in background, slice converted to copy
print(df)
```

```python
print(slice)
```
