<h3 align="left" style="color:blue">Pandas</h3>

```python
import pandas as pd
```

#### SERIES (one dimensional array)

```python
a = [1, 7, 2]
myvar = pd.Series(a)
print(myvar)                    #default labels starting index 0
```

```python
a = [1, 7, 2]
myvar = pd.Series(a,index=['a','b','c'])
print(myvar)
```

#### Series (key:value objects)

```python
calories = {"day1": 420, "day2": 380, "day3": 390}
myvar = pd.Series(calories)
print(myvar)                    # keys are labels/ index
```

```python
calories = {"day1": 420, "day2": 380, "day3": 390}
myvar = pd.Series(calories,index=['day1'])
print(myvar)                    # keys are labels/ index
```

#### DATAFRAMES (multi dimensional array)

#### 1. Create dataframe manually

```python
data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.loc[1])          # output is a panda series
```

```python
print(df.loc[[0,1]])
```

```python
df = pd.DataFrame(data, index = ["day1", "day2", "day3"])
print(df) 
```
