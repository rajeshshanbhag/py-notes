#### 2. Add column manually

#### 3. Add column using assign()

```python
data = {'Name': ['Pandas', 'Geeks', 'for', 'Geeks'],
        'Height': [1, 2, 3, 4],
        'Qualification': ['A', 'B', 'C', 'D']}
df_gfg = pd.DataFrame(data)
address = ['NewYork', 'Chicago', 'Boston', 'Miami']

df_gfg = df_gfg.assign(address = ['NewYork', 'Chicago', 'Boston', 'Miami'])           # alternate 1
print(df_gfg)
```

#### 4. Add column using map()

```python
data = {'Name': ['Pandas', 'Geeks', 'for', 'Geeks'],
        'Height': [1, 2, 3, 4],
        'Qualification': ['A', 'B', 'C', 'D']}
df_gfg = pd.DataFrame(data)
address = ['NewYork', 'Chicago', 'Boston', 'Miami']

address = {'Pandas': 'NewYork', 'Geeks': 'Chicago',                                   # alternate 2
            'for': 'Boston', 'Geeks_2': 'Miami'}
df_gfg['Address'] = df_gfg['Name'].map(address)
print(df_gfg)
```

#### Slicing columns

```python
df1 = pd.DataFrame({"a": [1, 2, 3, 4, 5, 6, 7], 
                    "b": [2, 3, 4, 2, 3, 4, 5], 
                    "c": [3, 4, 5, 2, 3, 4, 5], 
                    "d": [4, 5, 6, 2, 3, 4, 5], 
                    "e": [5, 6, 7, 2, 3, 4, 5]})
print(df1)
```

```python
df2 = df1.reindex(columns = ['c','b'])
print(df2)
```

```python
#df.loc[:,first:last:step]
df2 = df1.loc[:, "a":"e":2]
print(df2)
```

```python
#df.iloc[start:stop:step]
df2 = df1.iloc[:, 1:3:1]
print(df2)
```
