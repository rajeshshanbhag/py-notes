```python
import pandas as pd
pd.set_option("mode.copy_on_write", True)
```

#### Pivot Table

```python
sales = pd.DataFrame(
    data={
        "employee": [
            "Katrina",
            "Guanyu",
            "Jan",
            "Roman",
            "Jacqueline",
            "Paola",
            "Esperanza",
            "Alaina",
            "Egweyn",
        ],
        "sales": [14, 17, 6, 12, 8, 3, 7, 15, 5],
        "expertise": ["high", "high", "low", "low", "low", "low", "low", "high", "low"],
        "year": [2018, 2019, 2020, 2018, 2020, 2019, 2019, 2020, 2020],
    }
)
print(sales)
```

```python
avg_sales = sales.pivot_table(
    values="sales", index="year", columns="expertise", aggfunc="mean"
)
print(avg_sales)
```

#### Pivot

```python
df = pd.DataFrame(
    {
        "patient_name": ["Lekha", "Lekha", "Zhenzhen", "Zhenzhen", "Zhenzhen"],
        "reading": ["temp", "heart_rate", "temp", "heart_rate", "systolic_bp"],
        "reading_values": [101.5, 75, 98.3, 54, 115],
    }
)
print(df)
```

```python
df_pivoted = df.pivot(index="patient_name", columns="reading", values="reading_values").reset_index()
print(df_pivoted)
```

#### Melt

```python
df_melted = pd.melt(
    df_pivoted,
    id_vars=["patient_name"],
    value_vars=["heart_rate", "temp", "systolic_bp"],
)
print(df_melted)
```
