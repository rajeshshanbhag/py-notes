### Retirement Calculations

```python
# calculate retirement savings
#  start_age is the age (in months) when we start the calculation
#  initial is the initially saved balance at the start
#  working is a three tuple (months, contribution, rate of return)
#  retired is a three tuple (months, contribution, rate of return)
def calculateRetirement(start_age, initial, working, retired):
    for i in range(0,working[0]+1):
        if i==0:
            close_bal=initial
            age=start_age//12
            month=start_age%12
        else:
            interest=close_bal*working[2]
            close_bal=close_bal+interest+working[1]
            if month==11:
                month=0
                age+=1
            else:
                month+=1    
        print(f'Age {age:3d} month {month:2d} you have ${close_bal:.2f}')
    for i in range(0,retired[0]-1):
        interest=close_bal*retired[2]
        close_bal=close_bal+interest+retired[1]
        if month==11:
            month=0
            age+=1
        else:
            month+=1
        print(f'Age {age:3d} month {month:2d} you have ${close_bal:.2f}')
    pass
    

def main():
    working=(489,1000,0.045/12)
    retired=(384,-4000, 0.01/12)
    calculateRetirement(327, 21345, working, retired)
    pass

if __name__=="__main__":
    main()
    pass
```
