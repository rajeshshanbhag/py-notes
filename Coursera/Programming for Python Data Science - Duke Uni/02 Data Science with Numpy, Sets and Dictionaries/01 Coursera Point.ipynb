```python
For this assignment, you will write (in point.py) a class "Point",
which represents the x- and y-position of a point in the Cartesian
plane. As such, it should have fields for x and y.

The Point class should have the following attributes and methods:
  - A docstring summarizing the class.
  - __init__, which takes parameters x and y and default initializes x
    and y to 0.  
  - __str__ and __repr__, which give the string and representation of
    a Point object, e.g. for a Point with (1, -2), str would be 
    '(1, -2)', and repr would be 'Point(1, -2)'.  The __str__ and __repr__
    methods are standard things to write of an object, and get called
    when you use the str(x) and repr(x) functions.   Note that you
    usually call str(x) and repr(x) instead of x.__str__() or
    x.__repr__().  These two functions are basically simple
    wrappers for ease of use, i.e. you can think of them like this:
    def str(x):
      return x.__str__()
    def repr(x):
      return x.__repr__()
  - move, which takes parameters dx and dy and moves the x and y
    values of the point by those amounts.
  - distance_from, which takes a parameter other_pt and calculates its
    distance from the current point. Note: there are a variety of
    functions you might find useful in the math module.
```

```python
import math
from math import sqrt

class Point:
    # represents the x- and y-position of a point in the Cartesian plane.
    def __init__(self, x=0, y=0):
        self.x=x
        self.y=y
        pass
    def __str__(self):
        return f'{self.x,self.y}'
        #return ""
    def __repr__(self):
        return ("Point({}, {})".format(self.x, self.y))
    def move(self, dx, dy):
        self.x=dx+self.x
        self.y=dy+self.y
        pass
    def distance_from(self, other_pt):
        dist = sqrt((self.x-other_pt.x)**2 + (self.y-other_pt.y)**2)
        return dist
    pass

if __name__ == "__main__":
    p1=Point()
    p2=Point(3,4)
    p3=Point(-3,4)
    p4=Point(7,2)
    p5=Point(-9,16)
    points=[p1, p2, p3, p4, p5]
    for pa in points:
        print("---")
        print(pa)
        print(repr(pa))
        for pb in points:
            #print("pb",pb)
            print(pa.distance_from(pb))
            pass
        pass
    p1.move(1,1)
    p2.move(-2,3)
    p3.move(4,-5)
    for pa in points:
        print("---")
        print(pa)
        print(repr(pa))
        for pb in points:
            print(pa.distance_from(pb))
            pass
        pass
    pass
```
