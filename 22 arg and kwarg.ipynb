<h3 style="color:blue" align="left">ARG AND KWARG</h3>

#### *args argument

```python
def sum_all(*args):                                          # variable number of arguments
    total =0
    for n in args:
        total+=n
    return total

print(sum_all(1,2,3,4))
print(sum_all(1,2,3,4,5,6,7))
```

#### **kwargs argument

```python
def company_info(**kwargs):
    if 'ticker' in kwargs:
        print("Ticker: ", kwargs['ticker'])
    if 'ceo' in kwargs:
        print("CEO: ", kwargs['ceo'])
    if 'revenue' in kwargs:
        print("Revenue:", kwargs['revenue'])

company_info(ticker='AAPL', ceo="Tim Cook", revenue="200 billion")
```

```python
def company_info(**kwargs):
    for key in kwargs:
        print(key,kwargs[key])

company_info(ticker='AAPL', ceo="Tim Cook", revenue="200 billion",pe=20,pb=10.2)
```

#### *args amd **kwargs in combination

```python
def company_info(*args,**kwargs):
    print(f'I would like {args[0]} {kwargs['ticker']} shares')
```

```python
company_info(100,200,ticker='AAPL', ceo="Tim Cook")
```
