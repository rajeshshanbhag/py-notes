#### Access specific database

```python
mydb = mysql.connector.connect(
  host="localhost",
  user="root",
  password="root",
  database="mydatabase"
)

mycursor = mydb.cursor()
```

#### Create a Table

```python
#mycursor.execute("CREATE TABLE customers (name VARCHAR(255), address VARCHAR(255))")
mycursor.execute("CREATE TABLE customers (id INT AUTO_INCREMENT PRIMARY KEY, name VARCHAR(255), address VARCHAR(255))")
```

```python
mycursor.execute("SHOW TABLES")
for x in mycursor:
    print(x)
```

```python
mycursor.execute("ALTER TABLE customers ADD COLUMN id INT AUTO_INCREMENT PRIMARY KEY")
```

#### Insert record into table

```python
sql = "INSERT INTO customers (name, address) VALUES (%s, %s)"
val = ("John", "Highway 21")
mycursor.execute(sql, val)
mydb.commit()
print(mycursor.rowcount, "record inserted.")
```

#### Select Records

```python
mycursor.execute("SELECT * FROM customers")
#mycursor.execute("SELECT * FROM customers LIMIT 5")             # limit the result
#mycursor.execute("SELECT * FROM customers LIMIT 5 OFFSET 2")    # start select from offset record

myresult = mycursor.fetchall()

for x in myresult:
    print(x)
```

#### Select columns

```python
mycursor.execute("SELECT name, address FROM customers")

myresult = mycursor.fetchall()
#myresult = mycursor.fetchone()

for x in myresult:
    print(x)
```

#### Select with filter

```python
sql = "SELECT * FROM customers WHERE address ='Park Lane 38'"
mycursor.execute(sql)
myresult = mycursor.fetchall()
for x in myresult:
    print(x)
```

```python
sql = "SELECT * FROM customers WHERE address LIKE '%way%'"
mycursor.execute(sql)
myresult = mycursor.fetchall()
for x in myresult:
    print(x)
```

#### Prevent SQL Injection

```python
sql = "SELECT * FROM customers WHERE address = %s"
adr = ("Yellow Garden 2", )

#sql = "DELETE FROM customers WHERE address = %s"
#adr = ("Yellow Garden 2", )

#sql = "UPDATE customers SET address = %s WHERE address = %s"
#val = ("Valley 345", "Canyon 123")

mycursor.execute(sql, adr)
myresult = mycursor.fetchall()
for x in myresult:
    print(x)
```

#### Sort the Result

```python
sql = "SELECT * FROM customers ORDER BY name DESC"          # default = ASC
mycursor.execute(sql)
myresult = mycursor.fetchall()
for x in myresult:
    print(x)
```

#### Delete record

```python
sql = "DELETE FROM customers WHERE address = 'Mountain 21'"
mycursor.execute(sql)
mydb.commit()
print(mycursor.rowcount, "record(s) deleted")
```

#### Delete a table

```python
sql = "DROP TABLE IF EXISTS customers"
mycursor.execute(sql)
```

#### Update Table

```python
sql = "UPDATE customers SET address = 'Canyon 123' WHERE address = 'Valley 345'"
mycursor.execute(sql)
mydb.commit()
print(mycursor.rowcount, "record(s) affected")
```
