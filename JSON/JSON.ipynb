<h3 style="color:blue" align="left">JSON</h3>

```python
import json
```

#### JSON to dictionary

```python
# JSON string
json_string = '''
{
    "name": "Maulik",
    "age": 30,
    "is_employee": true,
    "skills": ["Python", "Data Analysis", "Machine Learning"],
    "address": {
        "street": "123 Main St",
        "city": "Wonderland"
    }
}
'''

record = json.loads(json_string)
record
```

```python
record['name']
```

#### Dictionary to JSON

```python
data = {
    "name": "Bob",
    "age": 25,
    "is_employee": False,
    "skills": ["JavaScript", "Web Development"],
    "address": {
        "street": "456 Elm St",
        "city": "Springfield"
    }
}

data_str = json.dumps(data)
data_str
```

#### write JSON to file

```python
with open('output.json', 'w') as file:
    json.dump(data, file, indent=4)
```

#### read JSON from file

```python
with open('products.json', 'r') as file:
    products = json.load(file)
type(products)
```
