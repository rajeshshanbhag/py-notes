#### Function as first class objects

```python
# Assigning functions to variables
def plus_one(number):
    return number + 1

add_one = plus_one                     # function assigned to variable, we can treat variable as a function
add_one(5)                             # variable calls the function
```

```python
# Passing functions as arguments to other functions
# Higher-order functions are functions that take one or more functions as arguments, return a function as a result or do both.
def plus_one(number):
    return number + 1

def function_call(function):
    number_to_add = 5
    return function(number_to_add)

function_call(plus_one)
```

```python
# Defining functions inside other functions
def plus_one(number):
    def add_one(number):
        return number + 1
    result = add_one(number)
    return result
plus_one(5)
```

#### Without decorator

```python
# Without decorator function
def decorator(func):
    def wrapper():                                   
        print("Before calling the function.")
        func()
        print("After calling the function.")
    return wrapper

def greet():
    print("Hello, World!")

decorate=decorator(greet)
decorate()
```

#### Decorator and Closure

```python
# When you create a decorator, the wrapper function (inside the decorator) is a closure.
# It retains access to the function being decorated and any additional state or arguments defined in the decorator function

# A simple decorator function
def decorator(func):                                           # calling a function
    def wrapper():                                             # if base function has arguments, wrapper should have (*args, **kwargs) as arguments
        print("Before calling the function.")
        func()
        print("After calling the function.")
    return wrapper                                             # returning a function with no parantheses (i.e. will not execute)

# Applying the decorator to a function
@decorator
def greet():
    print("Hello, World!")

greet()
```

#### Examples

```python
import time
def time_it(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args,**kwargs)
        end = time.time()
        print(func.__name__ +" took " + str((end-start)*1000) + "mil sec")
        return result
    return wrapper

@time_it
def calc_square(numbers):
    result = []
    for number in numbers:
        result.append(number*number)
    return result

@time_it
def calc_cube(numbers):
    result = []
    for number in numbers:
        result.append(number*number*number)
    return result

array = range(1,100000)
out_square = calc_square(array)
out_cube = calc_cube(array)
```
