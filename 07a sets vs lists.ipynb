<h3 align="left" style="color:blue">SETS vs LISTS</h3>

```python
import timeit

# Create a list and set with 10,000 elements
a = range(10000000)
s = set(a)

# Measure membership test times
t1= timeit.timeit(lambda: 9999999 in a, number=10000000)
t2 = timeit.timeit(lambda: 9999999 in s, number=10000000)

print(f"List time: {t1:.6f} | Set time: {t2:.6f}")
```
