```python
import uuid
```

#### uuid1() : Time and MAC Address

```python
print(uuid.uuid1())
```

#### Representations of uuid1()
* bytes : Returns id in form of 16 byte string.
* int : Returns id in form of 128-bit integer.
* hex : Returns random id as 32 character hexadecimal string

```python
id=uuid.uuid1()

# Representations of uuid1()
print ("The Representations of uuid1() are : ")
print ("byte Representation : ",end="")
print (repr(id.bytes))

print ("int Representation : ",end="")
print (id.int)

print ("hex Representation : ",end="")
print (id.hex)
```

#### Components of uuid1()
* version : version number of UUID.
* variant : The variant determining the internal layout of UUID.

```python
# Components of uuid1()
print ("The Components of uuid1() are : ")
print ("Version  : ",end="")
print (id.version)

print ("Variant : ",end="")
print (id.variant)
```

#### Fields of uuid1()
* time_low : The first 32 bits of id.
* time_mid : The next 16 bits of id.
* time_hi_version : The next 16 bits of id.
* clock_seq_hi_variant : Next 8 bits of id.
* clock_seq_low : Next 8 bits of id.
* node : Last 48 bits of id.
* time : Time component field of id.
* clock_seq : 14 bit sequence number.

```python
# Fields of uuid1()
print ("The Fields of uuid1() are : ")
print ("Fields  : ",end="")
print (id.fields)

print("\n")

# Time Component of uuid1()
print ("The time Component of uuid1() is : ")
print ("Time component  : ",end="")
print (id.node)
```

#### uuid4() : Random UUIDs

```python
print(uuid.uuid4())
```

#### Convert uuid to str()

```python
myuuid = uuid.uuid4()
print(myuuid)
myuuidStr = str(myuuid)                   
print(myuuidStr)                                        # i.e. normal UUID without the '-' characters
sameMyUuid = uuid.UUID(myuuidStr)
print(sameMyUuid)
assert myuuid == sameMyUuid
```

#### Using the UUID URN (Uniform Resource Name) Attribute

```python
my_uuid = uuid.uuid4()
 
uuid_urn = my_uuid.urn
print(uuid_urn) 
```

#### uuid3() / uuid5(): Deterministic UUIDs

```python
# Define a constant namespace UUID
namespace = uuid.UUID('905d5f1b-3c1f-407c-9c86-593b17f57323')
# Generate a UUID for a specific name within the namespace
print(uuid.uuid3(namespace, "tom"), "tom")
print(uuid.uuid3(namespace, "jerry"), "jerry")
print(uuid.uuid3(namespace, "tom"), "tom")
```
